# What does `@abstractmethod` do?

The decorator is implemented as
```python
def abstractmethod(func):
    func.__isabstractmethod__ = True
    return func
```

All it does on it's own. is set this attribute for tracking if some func is abstract

> Rem: how do decorators work?

This
```python
@abstractmethod
def pay(self, amount):
    pass
```

is same as

```python
def pay(self, amount):
    pass

pay = abstractmethod(pay)
```

````python
# testing the above
from abc import abstractmethod

@abstractmethod
def f():
    ...

f.__isabstractmethod__
````

# Then `ABC` providers the metaclass

`ABC` itself is roughly defined as

```python
class ABC(metaclass=ABCMeta):
    pass
```

> Rem: what does a metaclass do?

You always have some metaclass, the default one is "type". In general unless you are designing python language puzzles, all metaclassses you would want to make would inherit type. As such they add a little bit extra bells and whistles that you want over type.

What it helps in this context is that when you inherit from `ABC` you take over the `ABCMeta` as well which on _instantiation_ ( note this is why it won't throw if you just declare but don't create an object ) calls the `__new__` handling ( not really, it goes in C ) which checks if all there is an `__isabstractmethod__` remaining and throw if it is.
